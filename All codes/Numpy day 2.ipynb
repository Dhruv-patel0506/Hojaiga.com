```python
#can these arrays be concatenated???

import numpy as np

#2D NumPy array with shape (2, 3)
arr1 = np.array([[1, 2, 3],
                 [4, 5, 6]])

#2D NumPy array with shape (2, 2)
arr2 = np.array([[7, 8],
                 [9, 10]])

#2D NumPy array with shape (1, 3)
arr3 = np.array([[11, 12, 13]])
```

```python
np.concatenate((arr1,arr3),axis=0)
```

```python
np.concatenate((arr1,arr2),axis=1)
```

```python
#array 1 and 3 can be concatenated when we use axis=0
#array 1 and 2 can be concatenated when we use axis=1
```
